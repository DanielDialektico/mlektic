# Mlektic Phase 3 — Visual design system

This end-user notebook renders real Mlektic figures from fitted estimators. It demonstrates the additive visual system without unit-test assertions. The classic dashboard remains the default, and motion remains available in `dashboard`, `compact`, and `lesson` formats.

## Independent controls

- `theme` changes color, typography, line weight, and accessible redundancy.
- `format` changes composition and interaction structure.
- `density` controls mathematical context and aliases tabular `detail`.
- `size`, `width`, `height`, and `responsive` control the canvas.
- `reduced_motion=True` and `format="report"` are explicit static alternatives.

```python
import numpy as np
from IPython.display import Markdown, display
from sklearn.linear_model import LinearRegression, LogisticRegression

from mlektic import (
    available_themes,
    explain_logistic_prediction,
    explain_lr_prediction,
    get_theme_tokens,
    visualize_logistic,
    visualize_lr,
)

rng = np.random.default_rng(17)
X_linear = np.linspace(-3.0, 3.0, 70).reshape(-1, 1)
y_linear = 1.35 + 2.15 * X_linear[:, 0] + rng.normal(0.0, 0.28, len(X_linear))
linear_model = LinearRegression().fit(X_linear, y_linear)

X_logistic = np.linspace(-3.0, 3.0, 80).reshape(-1, 1)
latent_score = 1.8 * X_logistic[:, 0] - 0.25 + rng.normal(0.0, 0.7, len(X_logistic))
y_logistic = (latent_score > 0.0).astype(int)
logistic_model = LogisticRegression(random_state=17).fit(X_logistic, y_logistic)

available_themes()
```

## 1. Compatibility baseline

No Phase 3 option is required. This is the current classic animated dashboard.

```python
classic_figure = visualize_lr(
    linear_model,
    X_linear,
    y_linear,
    steps=24,
    max_frames=12,
    frame_duration=220,
    animation_mode="native",
)
display(classic_figure)
```

## 2. Academic theme, academic density, compact composition

The fitted-model derivation is separate from the evolving equation. The compact format reduces unused canvas space but preserves all 12 displayed animation states.

```python
academic_compact = visualize_lr(
    linear_model,
    X_linear,
    y_linear,
    steps=24,
    max_frames=12,
    frame_duration=220,
    animation_mode="native",
    theme="academic",
    format="compact",
    density="academic",
    size="notebook",
)
display(academic_compact)
```

## 3. Lesson format with fluid motion

The stage buttons isolate Data, Model, Objective, and Complete views. Select a stage and then use Play. The stages change visibility only; no mathematical state or animation frame is discarded.

```python
lesson_figure = visualize_lr(
    linear_model,
    X_linear,
    y_linear,
    steps=24,
    max_frames=12,
    frame_duration=240,
    animation_mode="native",
    theme="classroom",
    format="lesson",
    size="wide",
)
display(lesson_figure)
```

## 4. Accessible logistic animation

The accessible theme does not rely on color alone: observations use open markers, the model and objective use different line patterns, and decision boundaries receive a redundant dash style where applicable.

```python
accessible_logistic = visualize_logistic(
    logistic_model,
    X_logistic,
    y_logistic,
    steps=24,
    max_frames=12,
    frame_duration=240,
    show_loss=True,
    theme="accessible",
    size="notebook",
)
display(accessible_logistic)
```

## 5. Exact static report

Report format applies the exact last displayed state, removes temporal controls, and reclaims the slider region. It is intended for assignments, papers, and static documentation.

```python
report_figure = visualize_lr(
    linear_model,
    X_linear,
    y_linear,
    steps=24,
    max_frames=12,
    animation_mode="native",
    theme="academic",
    format="report",
    density="academic",
    size="notebook",
)
display(report_figure)
```

## 6. Prediction explainers use the same visual contract

```python
linear_prediction = explain_lr_prediction(
    linear_model,
    X_linear,
    y_linear,
    x_query=[[0.65]],
    theme="academic",
    format="compact",
)
display(linear_prediction)

logistic_prediction = explain_logistic_prediction(
    logistic_model,
    X_logistic,
    y_logistic,
    x_query=[[0.65]],
    theme="accessible",
)
display(logistic_prediction)
```

## 7. Responsive scaling and audit metadata

Responsive mode scales a selected composition; it does not claim CSS-style subplot reflow. Use another format when the information structure must change. Every resolved choice remains inspectable.

```python
responsive_figure = visualize_logistic(
    logistic_model,
    X_logistic,
    y_logistic,
    steps=18,
    max_frames=9,
    theme="academic",
    format="compact",
    responsive=True,
)
display(responsive_figure)
responsive_figure.layout.meta["mlektic_visual"]
```

## 9. Optional neural architecture view

This cell renders a real neural figure when the optional PyTorch dependency is installed. The neural classic design remains its baseline; shared themes are opt-in.

```python
try:
    import torch
    from mlektic import visualize_nn_architecture

    network = torch.nn.Sequential(
        torch.nn.Linear(2, 4),
        torch.nn.Tanh(),
        torch.nn.Linear(4, 1),
        torch.nn.Sigmoid(),
    )
    neural_figure = visualize_nn_architecture(
        network,
        torch.tensor([[0.3, -0.2]]),
        theme="academic",
        format="compact",
        responsive=True,
    )
    display(neural_figure)
except ImportError:
    display(Markdown("Install the optional dependency with `pip install mlektic[torch]` to render this figure."))
```
