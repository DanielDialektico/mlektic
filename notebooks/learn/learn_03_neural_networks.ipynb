## What, why, and how the network learns

Four noisy XOR clouds need a curved boundary. The next four views use the same recorded run.

```python
from mlektic import visualize_nn_decision, visualize_nn_gradient_flow, visualize_nn_representation, visualize_nn_training_step
from notebooks._support import torch_depth_case, torch_xor_blobs_case
blob_model,blob_X,blob_y,blob_history,blob_loss=torch_xor_blobs_case()
```

### `LEARN-NN-DECISION`

**Inspect:** each hidden unit folds the plane along z_j=0 and the readout weights combine the folds into the boundary

```python
case_heading("LEARN-NN-DECISION", "each hidden unit folds the plane along z_j=0 and the readout weights combine the folds into the boundary")
display(visualize_nn_decision(blob_model,blob_X,blob_history,targets=blob_y))
```

### `LEARN-NN-REPRESENTATION`

**Inspect:** the affine map keeps lines straight, tanh bends the plane, and the boundary becomes a vertical line

```python
case_heading("LEARN-NN-REPRESENTATION", "the affine map keeps lines straight, tanh bends the plane, and the boundary becomes a vertical line")
display(visualize_nn_representation(blob_model,blob_X,blob_history,targets=blob_y))
```

### `LEARN-NN-TRAINING-STEP`

**Inspect:** forward pulses, per-sample loss, deltas travelling back, edge gradients, and one update

```python
case_heading("LEARN-NN-TRAINING-STEP", "forward pulses, per-sample loss, deltas travelling back, edge gradients, and one update")
display(visualize_nn_training_step(blob_model,blob_X[5],blob_y[5],blob_loss,blob_history,checkpoint=1))
```

### `LEARN-NN-GRADIENT-FLOW`

**Inspect:** the first sigmoid layer receives orders of magnitude less gradient than the ReLU one

```python
case_heading("LEARN-NN-GRADIENT-FLOW", "the first sigmoid layer receives orders of magnitude less gradient than the ReLU one")
_,sigmoid_history=torch_depth_case('sigmoid')
_,relu_history=torch_depth_case('relu')
display(visualize_nn_gradient_flow({'Sigmoid':sigmoid_history,'ReLU':relu_history}))
```
