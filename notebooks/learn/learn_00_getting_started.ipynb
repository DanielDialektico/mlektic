## 1. Fit a model

The estimator is already fitted before Mlektic receives it.

```python
from sklearn.linear_model import LinearRegression
from mlektic import explain_lr_prediction, visualize_lr
X=np.linspace(-2,2,60).reshape(-1,1)
y=1.2+2.0*X[:,0]
model=LinearRegression().fit(X,y)
```

### `LEARN-GETTING-STARTED`

**Inspect:** synthetic interpolation is labeled and ends at the fitted model

```python
case_heading("LEARN-GETTING-STARTED", "synthetic interpolation is labeled and ends at the fitted model")
display(visualize_lr(model,X,y,steps=12,max_frames=7,format='lesson',theme='classroom'))
```
